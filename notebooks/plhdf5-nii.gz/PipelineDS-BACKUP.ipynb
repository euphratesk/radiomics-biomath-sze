```python
from radiomics import featureextractor
extractor = featureextractor.RadiomicsFeatureExtractor()
```

```python
import pandas as pd
import numpy as np
```

```python
kits_path = "nii"

for case in os.listdir(kits_path):

    case_dir = os.path.join(kits_path, case)

    image_path = os.path.join(case_dir, "imaging.nii.gz")
    mask_path = os.path.join(case_dir, "segmentation.nii.gz")

    image = sitk.ReadImage(image_path)
    mask = sitk.ReadImage(mask_path)

    features = extractor.execute(image, mask)

    row = dict(features)
    row["case_id"] = case
    row["dataset"] = "kits"
```

. Diagnostic Columns (Metadata)

Columns beginning with:

diagnostics_Versions_*
diagnostics_Configuration_*
diagnostics_Image-original_*
diagnostics_Mask-original_*

Definition
These columns store information about the radiomics extraction process and the image properties. They are not radiomics features used for modeling.

Examples and meaning

diagnostics_Versions_PyRadiomics
Version of the PyRadiomics library used for feature extraction.

diagnostics_Versions_Numpy / SimpleITK / PyWavelet / Python
Versions of supporting libraries used in the pipeline.

diagnostics_Configuration_Settings
The parameters used during feature extraction.

diagnostics_Configuration_EnabledImageTypes
Which image transformations (original, wavelet, LoG) were enabled.

diagnostics_Image-original_Spacing
Voxel spacing of the CT image (millimeters).

diagnostics_Image-original_Size
Dimensions of the CT image.

diagnostics_Image-original_Mean / Minimum / Maximum
Basic statistics of CT intensity values.

diagnostics_Mask-original_VoxelNum
Number of voxels in the segmented tumor.

diagnostics_Mask-original_BoundingBox
Spatial limits of the tumor region.

diagnostics_Mask-original_CenterOfMass
Geometric center of the tumor mask.

***
Contribution to dataset
These columns are mainly used for:

• reproducibility of experiments
• quality control
• verifying CT image properties

They are usually removed before machine learning modeling.
***
2. Shape Features

Columns beginning with:

original_shape_*

Definition
These features describe the three-dimensional geometry of the tumor segmentation.

Examples

original_shape_Elongation
Measures how stretched the tumor shape is.

original_shape_Flatness
Indicates whether the tumor is flat or spherical.

original_shape_MajorAxisLength
Length of the main axis of the tumor.

original_shape_MinorAxisLength
Second principal axis of the tumor.

original_shape_Maximum3DDiameter
Largest distance between two tumor points.

original_shape_SurfaceArea
Surface area of the tumor.

original_shape_VoxelVolume
Tumor volume measured in voxels.

original_shape_Sphericity
How closely the tumor shape resembles a sphere.

Contribution to dataset
Shape features represent tumor morphology such as size, compactness, and irregularity.

These features can help differentiate tumor types because malignant tumors often have more irregular shapes.

3. First-Order Statistics (Intensity Features)

Columns beginning with:

original_firstorder_*

Definition
These features describe the statistical distribution of CT intensity values (Hounsfield Units) inside the tumor.

Examples

original_firstorder_Mean
Average CT intensity of the tumor.

original_firstorder_Median
Median HU value.

original_firstorder_Minimum
Lowest HU value inside the tumor.

original_firstorder_Maximum
Highest HU value inside the tumor.

original_firstorder_Variance
Variance of intensity values.

original_firstorder_Skewness
Asymmetry of the HU distribution.

original_firstorder_Kurtosis
Sharpness of the HU distribution peak.

original_firstorder_Entropy
Randomness of intensity values.

Contribution to dataset
These features represent tumor density and tissue composition.

They are useful for detecting fat, necrosis, or contrast enhancement patterns.

4. GLCM Features (Gray Level Co-occurrence Matrix)

Columns beginning with:

original_glcm_*

Definition
GLCM features describe how often pairs of intensity values occur next to each other in the image.

Examples

original_glcm_Contrast
Measures local intensity differences.

original_glcm_Correlation
Measures correlation between neighboring voxels.

original_glcm_Entropy
Measures randomness of texture.

original_glcm_Idm (Inverse Difference Moment)
Measures image smoothness.

Contribution to dataset
GLCM features capture tumor texture patterns and heterogeneity.

5. GLDM Features (Gray Level Dependence Matrix)

Columns beginning with:

original_gldm_*

Definition
These features measure how many neighboring voxels depend on a central voxel with similar intensity.

Examples

original_gldm_DependenceEntropy
Complexity of voxel dependence patterns.

original_gldm_DependenceVariance
Variation in dependence size.

original_gldm_LargeDependenceEmphasis
Presence of large homogeneous structures.

Contribution to dataset
These features describe local homogeneity and spatial dependence of intensities.

6. GLRLM Features (Gray Level Run Length Matrix)

Columns beginning with:

original_glrlm_*

Definition
These features analyze consecutive sequences (runs) of voxels that have the same intensity value.

Examples

original_glrlm_ShortRunEmphasis
Presence of short intensity runs.

original_glrlm_LongRunEmphasis
Presence of long homogeneous runs.

original_glrlm_RunEntropy
Randomness of run lengths.

Contribution to dataset
GLRLM features capture directional texture patterns and structural continuity.

7. GLSZM Features (Gray Level Size Zone Matrix)

Columns beginning with:

original_glszm_*

Definition
These features measure connected regions (zones) that share the same gray-level intensity.

Examples

original_glszm_SmallAreaEmphasis
Presence of small homogeneous zones.

original_glszm_LargeAreaEmphasis
Presence of large homogeneous zones.

original_glszm_ZoneEntropy
Randomness of zone distribution.

Contribution to dataset
GLSZM features describe spatial clustering of intensities within tumors.

8. NGTDM Features (Neighborhood Gray Tone Difference Matrix)

Columns beginning with:

original_ngtdm_*

Definition
These features compare a voxel intensity with the average intensity of its neighboring voxels.

Examples

original_ngtdm_Coarseness
Measures texture smoothness.

original_ngtdm_Busyness
Measures rapid intensity changes.

original_ngtdm_Contrast
Measures local contrast differences.

Contribution to dataset
These features represent texture granularity and local intensity variation.

9. Dataset Identification Columns

case_id
Unique identifier for each patient or scan.

dataset
Indicates which dataset the sample belongs to (for example HDF5 or NIfTI dataset).

Contribution to dataset
These columns allow tracking of cases and detecting potential dataset bias.

Overall Concept

Your radiomics dataset contains five main types of information:

Metadata (diagnostics columns)
Tumor shape features
Intensity statistics
Texture features
Dataset identifiers

Machine learning models usually use:

shape + intensity + texture features

while diagnostic metadata is removed.
