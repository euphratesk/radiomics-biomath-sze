```python
import h5py
import numpy as np
import pandas as pd
import os
import SimpleITK as sitk
import csv
from radiomics import featureextractor
extractor = featureextractor.RadiomicsFeatureExtractor()
```

hdf5_folder = "/media/euphrates/DS-Backup/angiomyolipoma/ucsf-rmac-dataset"
rd_mics_hdf5 = "/media/euphrates/DS-Backup/csv_aml_ucsf_rmac"


```python
for file in os.listdir(hdf5_folder):

    input_path = os.path.join(hdf5_folder, file)
    output_path = os.path.join(rd_mics_hdf5, file.replace("hdf5","csv"))

    with h5py.File(input_path, "r") as f:

        image_array = f["portven"][:]
        mask_array = f["mask"][:]

        image = sitk.GetImageFromArray(image_array)
        mask = sitk.GetImageFromArray(mask_array)

        features = extractor.execute(image, mask)

        row_h= dict(features)
        row_h["case_id"] = file
    with open(output_path,w) as f:
        w = csv.writer(f)
        w.writerows(row_h.items())
```
